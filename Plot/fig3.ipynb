```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde

# 假设 data 是你已经读取好的 ndarray，形状为 (num_samples, num_phi)
# 如果你已经有了 data，就不用重复定义

num_samples, num_phi = data.shape
phi = np.arange(num_phi)

# 展开为所有样本的 (phi, f) 坐标
phi_all = np.tile(phi, num_samples)
f_all = data.flatten()

# ===== 图2：2D 直方图（像你的图） =====
# plt.subplot(1, 2, 2)
hist, xedges, yedges = np.histogram2d(phi_all, f_all, bins=[num_phi, 200])
plt.figure(figsize=(8, 4))
plt.imshow(hist.T, origin='lower', aspect='auto', cmap='viridis',extent=[0,360,f_all.min()/np.pi*180,f_all.max()/np.pi*180])
plt.colorbar().set_label(label='Counts',rotation=270,labelpad=15)
plt.xlabel('phi')
plt.ylabel('EVPA')
plt.title('EVPA Histogram')

plt.tight_layout()
plt.savefig('fig3_EVPA_GRMHD_Histogram',  bbox_inches='tight')
plt.show()
```
